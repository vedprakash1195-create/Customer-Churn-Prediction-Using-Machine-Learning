```python

# Remove customerID and recreate X and y
df = df.drop(columns=["customerID"], errors="ignore")

X = df.drop(columns=["Churn"])
y = df["Churn"]

numeric_cols = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_cols = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("customerID in X:", "customerID" in X.columns)
print("Total features:", X.shape[1])
```

```python

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Numeric columns
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical columns
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both pipelines
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

print("Preprocessor ready!")
print("Numeric columns:", len(numeric_cols))
print("Categorical columns:", len(categorical_cols))

```

```python

from sklearn.base import clone
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Logistic Regression
model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

# Decision Tree
dt_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", DecisionTreeClassifier(
        max_depth=6,
        class_weight="balanced",
        random_state=42
    ))
])

# Random Forest
rf_model = Pipeline([
    ("preprocessor", clone(preprocessor)),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# Train all models
model.fit(X_train, y_train)
dt_model.fit(X_train, y_train)
rf_model.fit(X_train, y_train)

print("All three models trained successfully!")
```

```python

from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
import pandas as pd

# Split training data into fit and validation sets
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Train a fresh Decision Tree on the fit set
tuning_model = clone(dt_model)
tuning_model.fit(X_fit, y_fit)

# Get churn probabilities for validation data
y_prob_val = tuning_model.predict_proba(X_val)[:, 1]

print("Fit rows:", X_fit.shape[0])
print("Validation rows:", X_val.shape[0])
print("Ready for threshold tuning!")
```

```python

import pandas as pd
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score

# Get churn probabilities on validation data
y_prob_val = tuning_model.predict_proba(X_val)[:, 1]

thresholds = [0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60, 0.65]

threshold_results = []

for threshold in thresholds:
    y_pred_threshold = (y_prob_val >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val, y_pred_threshold, labels=[0, 1]
    ).ravel()

    threshold_results.append({
        "Threshold": threshold,
        "Precision": round(precision_score(y_val, y_pred_threshold, zero_division=0), 3),
        "Recall": round(recall_score(y_val, y_pred_threshold, zero_division=0), 3),
        "F1 Score": round(f1_score(y_val, y_pred_threshold, zero_division=0), 3),
        "False Positives": int(fp),
        "False Negatives": int(fn),
        "True Positives": int(tp)
    })

threshold_df = pd.DataFrame(threshold_results)
display(threshold_df)
```

```python

eligible = threshold_df[threshold_df["Recall"] >= 0.75].copy()

if not eligible.empty:
    eligible = eligible.sort_values(
        by=["Precision", "F1 Score"],
        ascending=False
    )
    display(eligible)

    best_threshold = eligible.iloc[0]["Threshold"]
    print("Selected threshold:", best_threshold)
else:
    print("No threshold meets the 75% recall target.")
```

```python

from sklearn.metrics import confusion_matrix, classification_report
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Predict churn probabilities on test data
y_prob_test = tuning_model.predict_proba(X_test)[:, 1]

# Apply selected threshold
y_pred_test = (y_prob_test >= best_threshold).astype(int)

# Evaluation
tn, fp, fn, tp = confusion_matrix(
    y_test, y_pred_test, labels=[0, 1]
).ravel()

print("Selected Threshold:", best_threshold)
print("Accuracy:", round(accuracy_score(y_test, y_pred_test), 3))
print("Precision:", round(precision_score(y_test, y_pred_test, zero_division=0), 3))
print("Recall:", round(recall_score(y_test, y_pred_test, zero_division=0), 3))
print("F1 Score:", round(f1_score(y_test, y_pred_test, zero_division=0), 3))

print("\nConfusion Matrix:")
print([[tn, fp], [fn, tp]])

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_test,
    target_names=["No Churn", "Churn"],
    zero_division=0
))
```
